```python
from collections import Counter
data = [
    ["Sunny", "No"],
    ["Sunny", "No"],
    ["Cloudy", "Yes"],
    ["Rain", "Yes"],
    ["Rain", "No"]
]
def majority_class(data):
    labels = [row[1] for row in data]
    return Counter(labels).most_common(1)[0][0]
result = majority_class(data)
print("Majority Class:", result)
```
